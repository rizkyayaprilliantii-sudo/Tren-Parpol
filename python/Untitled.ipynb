```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sqlalchemy import create_engine, text
from sklearn.cluster import KMeans


DB_URI = "mysql+pymysql://root:@127.0.0.1:3306/trenparpol"

engine = create_engine(DB_URI)

print("📊 Mengambil data rekap suara 2024...")

df = pd.read_sql("SELECT id_partai, total_suara FROM rekap_suara_2024", engine)

print("Data yang akan di-clustering:")
display(df)

X = df[['total_suara']]

kmeans = KMeans(n_clusters=3, random_state=42, n_init='auto')
df['cluster'] = kmeans.fit_predict(X)

centroids = kmeans.cluster_centers_

df['nilai_centroid'] = df['cluster'].map({i: centroids[i][0] for i in range(3)})

df['jarak_centroid'] = abs(df['total_suara'] - df['nilai_centroid'])

sorted_centroids = np.sort(centroids.flatten())

kategori_map = {
    np.where(centroids.flatten() == sorted_centroids[0])[0][0]: 'Rendah',
    np.where(centroids.flatten() == sorted_centroids[1])[0][0]: 'Sedang',
    np.where(centroids.flatten() == sorted_centroids[2])[0][0]: 'Tinggi'
}

df['kategori_cluster'] = df['cluster'].map(kategori_map)


print("\n HASIL CLUSTERING K-MEANS 2024:")
df_sorted = df.sort_values(by='total_suara', ascending=False)
display(df_sorted)

print("\n💾 Menyimpan hasil ke tabel 'hasil_cluster_2024'...")

with engine.begin() as conn:
    conn.execute(text("TRUNCATE TABLE hasil_cluster_2024"))

df.to_sql('hasil_cluster_2024', engine, if_exists='append', index=False)

print("Data clustering 2024 berhasil disimpan.")
```

```python
import pandas as pd
import numpy as np
from sqlalchemy import create_engine, text
from sklearn.cluster import KMeans

DB_URI = "mysql+pymysql://root:@127.0.0.1:3306/trenparpol"
engine = create_engine(DB_URI)

print("📊 Mengambil data rekap suara 2019...")

df_2019 = pd.read_sql("SELECT id_partai, total_suara FROM rekap_suara_2019", engine)

print("Data yang akan di-clustering:")
display(df_2019)

X_2019 = df_2019[['total_suara']]

kmeans_2019 = KMeans(n_clusters=3, random_state=42, n_init='auto')
df_2019['cluster'] = kmeans_2019.fit_predict(X_2019)


centroids_2019 = kmeans_2019.cluster_centers_
df_2019['nilai_centroid'] = df_2019['cluster'].map({i: centroids_2019[i][0] for i in range(3)})
df_2019['jarak_centroid'] = abs(df_2019['total_suara'] - df_2019['nilai_centroid'])


sorted_centroids_2019 = np.sort(centroids_2019.flatten())

kategori_map_2019 = {
    np.where(centroids_2019.flatten() == sorted_centroids_2019[0])[0][0]: 'Rendah',
    np.where(centroids_2019.flatten() == sorted_centroids_2019[1])[0][0]: 'Sedang',
    np.where(centroids_2019.flatten() == sorted_centroids_2019[2])[0][0]: 'Tinggi'
}

df_2019['kategori_cluster'] = df_2019['cluster'].map(kategori_map_2019)

print("\n HASIL CLUSTERING K-MEANS 2019:")
df_2019_sorted = df_2019.sort_values(by='total_suara', ascending=False)
display(df_2019_sorted)

print("\n💾 Menyimpan hasil ke tabel 'hasil_cluster_2019'...")

with engine.begin() as conn:
    conn.execute(text("TRUNCATE TABLE hasil_cluster_2019"))

df_2019.to_sql('hasil_cluster_2019', engine, if_exists='append', index=False)

print("Data clustering 2019 berhasil disimpan.")
```

```python
print("Mengecek perbedaan partai 2019 dan 2024...")

partai_2019 = set(df_2019['id_partai'])
partai_2024 = set(df['id_partai'])

hanya_2019 = sorted(partai_2019 - partai_2024)
hanya_2024 = sorted(partai_2024 - partai_2019)
ada_keduanya = sorted(partai_2019 & partai_2024)

print("\nPartai yang ada di kedua periode:")
print(ada_keduanya)

print("\nPartai yang hanya ada di 2019:")
print(hanya_2019)

print("\nPartai yang hanya ada di 2024:")
print(hanya_2024)

print("\nJumlah:")
print("Ada di kedua periode :", len(ada_keduanya))
print("Hanya 2019          :", len(hanya_2019))
print("Hanya 2024          :", len(hanya_2024))
```

```python
print("Membuat perbandingan suara 2019 dan 2024...")

perbandingan = pd.merge(
    df_2019[['id_partai', 'total_suara']],
    df[['id_partai', 'total_suara']],
    on='id_partai',
    how='inner',
    suffixes=('_2019', '_2024')
)

perbandingan['selisih_suara'] = (
    perbandingan['total_suara_2024']
    - perbandingan['total_suara_2019']
)

perbandingan['persentase_perubahan'] = (
    perbandingan['selisih_suara']
    / perbandingan['total_suara_2019']
    * 100
)

perbandingan = perbandingan.sort_values(
    by='total_suara_2024',
    ascending=False
)

print("\nHASIL PERBANDINGAN 14 PARTAI:")
display(perbandingan)
```

```python
print("Menggabungkan hasil clustering 2019 dan 2024...")

cluster_2019 = df_2019[
    ['id_partai', 'cluster', 'kategori_cluster']
].rename(columns={
    'cluster': 'cluster_2019',
    'kategori_cluster': 'kategori_cluster_2019'
})

cluster_2024 = df[
    ['id_partai', 'cluster', 'kategori_cluster']
].copy()

cluster_2024 = cluster_2024.rename(columns={
    'cluster': 'cluster_2024',
    'kategori_cluster': 'kategori_cluster_2024'
})

perbandingan_cluster = perbandingan.merge(
    cluster_2019,
    on='id_partai',
    how='left'
)

perbandingan_cluster = perbandingan_cluster.merge(
    cluster_2024,
    on='id_partai',
    how='left'
)

print("\nHASIL PERBANDINGAN + CLUSTER:")
display(perbandingan_cluster)
```

```python
print("Menentukan perubahan cluster 2019 → 2024...")

def status_cluster(row):
    if row['kategori_cluster_2019'] == row['kategori_cluster_2024']:
        return 'Tetap'
    else:
        return (
            row['kategori_cluster_2019']
            + ' → '
            + row['kategori_cluster_2024']
        )

perbandingan_cluster['perubahan_cluster'] = (
    perbandingan_cluster.apply(status_cluster, axis=1)
)

print("\nHASIL AKHIR PERBANDINGAN CLUSTER:")
display(perbandingan_cluster)
```
